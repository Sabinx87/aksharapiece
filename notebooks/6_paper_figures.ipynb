# AksharaPiece paper figures

Makes the 6 paper figures (PNG and PDF) and the tables from the saved results. The code and result files are inside this notebook, so nothing needs to be attached.

**Settings:** Accelerator **None** (no GPU needed). Internet can stay off.

Click **Run All**, or **Save Version → Save & Run All** to keep the output. The figures appear below and in the **Output** panel as `aksharapiece_figures_output.zip`.

```python
import base64, io, os, subprocess, sys, zipfile

WORK = "/kaggle/working" if os.path.isdir("/kaggle/working") else os.getcwd()
KIT_B64 = ""
zipfile.ZipFile(io.BytesIO(base64.b64decode(KIT_B64))).extractall("/tmp")
os.chdir("/tmp/aksharapiece_figures")
print(open("README.md").read())
```

```python
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "matplotlib", "numpy", "scipy"], check=False)
r = subprocess.run([sys.executable, "scripts/make_figures.py", "--out", "figures"], capture_output=True, text=True)
print(r.stdout[-3000:], r.stderr[-3000:])
```

## Figures

```python
from IPython.display import Image, Markdown, display
import glob
for f in sorted(glob.glob("figures/*.png")):
    display(Markdown(f"**{os.path.basename(f)}**"))
    display(Image(f))
```
